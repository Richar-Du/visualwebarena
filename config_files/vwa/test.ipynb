```python
import json
import os
import random

tasks = ['classifieds', 'reddit', 'shopping']

tasks = ['shopping']

def if_url_related(filename, task):
    d = json.load(open(os.path.join(f"/home/liuzikang/visualwebarena/config_files/vwa/test_{task}",filename), 'r'))
    if 'reference_answer_raw_annotation' in d['eval'].keys() and ('.jpg' in d['eval']['reference_answer_raw_annotation'] or 'url' in d['intent']):
        return True
    return False
```

```python
if_url_related('204.json', 'shopping')
```

```python
task = 'shopping'
all_configs = os.listdir(f"/home/liuzikang/visualwebarena/config_files/vwa/test_{task}")

cnt = 0

for ac in all_configs:
    cnt += if_url_related(ac, task)
    
cnt 
```

```python
config_subset_dict = {}
for task in tasks:
    k = 20
    all_configs = os.listdir(f"/home/liuzikang/visualwebarena/config_files/vwa/test_{task}")
    extract_configs = [ac for ac in all_configs if not if_url_related(ac, task)]
    config_subset_dict[task] = random.sample(all_configs, k)
```

```python
json.dump(config_subset_dict, open('config_subset_shopping_20.json', 'w'), indent=4)
```
